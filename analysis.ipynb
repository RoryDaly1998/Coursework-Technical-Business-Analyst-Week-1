```python
import pandas as pd

data = pd.read_csv("recovery_activity_tracker.csv")
```

```python
summary = data.groupby('activity_type')['minutes_spent'].agg(['sum'])
#summary.columns = ['Total Minutes', 'Avg Minutes per Task', 'Task Count']
#summary['Avg Minutes per Task'] = summary['Avg Minutes per Task'].round(2)
summary
```

```python
summary = data.groupby('outcome_code')['minutes_spent'].agg(['sum', 'mean', 'count'])
summary.columns = ['Total Minutes', 'Avg Minutes per Task', 'Task Count']
summary['Avg Minutes per Task'] = summary['Avg Minutes per Task'].round(2)
summary
```

```python
total_minutes = data['minutes_spent'].sum()
num_accounts = data['account_id'].nunique()
overall_avg_per_account = total_minutes / num_accounts
overall_avg_per_account
```
